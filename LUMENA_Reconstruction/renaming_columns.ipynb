```python
import pandas as pd
```

```python
tiles_df = pd.read_csv("path/to/QuPath_tile_df.csv",index_col=None)
#anno_df = pd.read_csv("path/to/QuPath_annotation_df.csv",index_col=None) #can rename annotation df column headers if needed
```

```python
print(tiles_df.columns) # check the naming structure of the column headers to identify what to change
```

```python
#tiles_df = tiles_df.rename(columns={'Tissue.ID':'Parent'}) # fully rename the specified column
# examples of replacing parts of the column names with other text (e.g., replace spaces with underscores, % sign with 'percent', etc.)
tiles_df.columns = tiles_df.columns.str.replace(' ','_') 
tiles_df.columns = tiles_df.columns.str.replace(':','')
tiles_df.columns = tiles_df.columns.str.replace('%','Percent')
tiles_df.columns = tiles_df.columns.str.replace('µm^2','um_2')
tiles_df.columns = tiles_df.columns.str.replace('µm','um')
tiles_df = tiles_df.add_prefix('Tiles_') #Adds 'Tiles_' to the beginning of every column name
```

```python
print(tiles_df.columns) #check new names for consistency and/or other things to change in the column names
```

```python
tiles_df = tiles_df.rename(columns={'Tiles_Unnamed: 0':'Index'}) #if an unnamed column (usually an indexing type column) is included, rename to 'Index' to exclude it when saving new df
```

```python
tiles_df.head() #check the column names and the first 5 rows of data
```
